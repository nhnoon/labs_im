# Geometric & Intensity Transformations


Apply classical image processing operations using Python and the `(PIL)` library. Complete both tasks and save your output images for comparison.


```python
# Import the libraries
from PIL import Image, ImageOps
import numpy as np
from IPython.display import display
import base64
import io
```

```python
# Setup: load the embedded lab image (kept self-contained)
from io import BytesIO
import base64
_INPUT_IMAGE_B64 = ""
img = Image.open(BytesIO(base64.b64decode(_INPUT_IMAGE_B64))).convert("RGB")
display(img)
```

### 1. Scale (increase size)
Double the image dimensions using `Image.resize()` with high-quality resampling `(LANCZOS)`.

```python
# Get the size of the image
width, height = img.size

# scaling factors
cx = 2
cy = 2

new_width = int(width * cx)
new_height = int(height * cy)

print(f"Original size: {width} × {height}")
print(f"Target size:   {new_width} × {new_height}")
```

```python
# ── 1. Increase size (scale ×2) ────────────────────

# Resize the image using PIL's built-in method
scaled_img = img.resize(
    (new_width, new_height),
    resample=Image.Resampling.LANCZOS
)

display(scaled_img)
```

```python
# Save the scaled image and print the sizes
# (The new image name should be "task1_1_scaled.jpg")

scaled_img.save("task1_1_scaled.jpg", quality=95)

print("Saved: task1_1_scaled.jpg")
print(f"Original size: {img.size}")
print(f"Scaled size:   {scaled_img.size}")
```

 non-uniform scale (cx=2, cy=1) → stretch horizontally only

```python
# Non-uniform scaling factors
cx_nonuniform = 2
cy_nonuniform = 1

nonuniform_size = (
    int(width * cx_nonuniform),
    int(height * cy_nonuniform)
)

print(f"Non-uniform scale factors: cx={cx_nonuniform}, cy={cy_nonuniform}")
print(f"Target size: {nonuniform_size}")
```

```python
# Stretch horizontally only using high-quality LANCZOS resampling
nonuniform_scaled_img = img.resize(
    nonuniform_size,
    resample=Image.Resampling.LANCZOS
)

display(nonuniform_scaled_img)
```

```python
# Save the non-uniformly scaled result
nonuniform_scaled_img.save("task1_1_nonuniform_scaled.jpg", quality=95)

print("Saved: task1_1_nonuniform_scaled.jpg")
print(f"Original size:            {img.size}")
print(f"Non-uniform scaled size:  {nonuniform_scaled_img.size}")
```

### 2. Rotate 120°
Rotate the image by 120 degrees, expanding the canvas to fit the full rotated image.

```python
# ── 2. Rotate 120 degrees ──────────────────────────
# Rotate the image and expand the canvas to keep the full result.

rotated_img = img.rotate(
    120,
    resample=Image.Resampling.BICUBIC,
    expand=True,
    fillcolor=0
)

# Save the rotated image
# (The new image name should be "task1_2_rotated.jpg")
rotated_img.save("task1_2_rotated.jpg", quality=95)

print("Saved: task1_2_rotated.jpg")
print(f"Original size: {img.size}")
print(f"Rotated canvas size: {rotated_img.size}")
display(rotated_img)
```

### 3. Shear

```python
# -- c. Get the image dimensions ────────────────────
width, height = img.size
print(f"Image dimensions: {width} × {height}")
```

```python
# -- d. define the shear matrix ──────────────────────────
# Choose X-axis shear. The shear factor controls how much the image slants.
shear_x = 0.5
shear_y = 0.0

# For positive X-shear: x' = x + shear_x * y
# Increase the output canvas width so the full image is preserved.
sheared_width = int(round(width + abs(shear_x) * height))
sheared_height = height

print(f"X-axis shear factor: {shear_x}")
print(f"Output canvas: {sheared_width} × {sheared_height}")
```

```python
# -- e. Apply the shear transformation to the image ──────────────────────────
# PIL's transform() uses the inverse mapping from output coordinates
# back to input coordinates.
#
# Forward X-shear:
# x' = x + shx*y
# y' = y
#
# Inverse mapping used by PIL:
# x = x' - shx*y'
# y = y'

sheared_img = img.transform(
    (sheared_width, sheared_height),
    Image.Transform.AFFINE,
    (1, -shear_x, 0,
     0,  1,       0),
    resample=Image.Resampling.BICUBIC,
    fillcolor=0
)

display(sheared_img)
```

```python
# -- f. Save the sheared image
# (The new image name should be "task1_3_sheared.jpg")

sheared_img.save("task1_3_sheared.jpg", quality=95)

print("Saved: task1_3_sheared.jpg")
print(f"Original size: {img.size}")
print(f"Sheared size:  {sheared_img.size}")
```

### Experiment and compare
Try these:

— Change shear factor from 0.5 to 0.1, 0.3, 0.8 and compare results

— Switch from X-axis to Y-axis shear matrix

— Update the canvas multiplier to match your new factor

— Try combining X and Y shear in one matrix

```python
# Experiment with different shear factors and directions

def shear_image(image, shx=0.0, shy=0.0):
    w, h = image.size
    det = 1 - (shx * shy)
    if abs(det) < 1e-12:
        raise ValueError("Invalid shear combination: transformation matrix is singular.")

    # Canvas large enough for positive shear factors used in this experiment
    out_w = int(round(w + abs(shx) * h))
    out_h = int(round(h + abs(shy) * w))

    # Inverse of [[1, shx], [shy, 1]]
    a = 1 / det
    b = -shx / det
    d = -shy / det
    e = 1 / det

    return image.transform(
        (out_w, out_h),
        Image.Transform.AFFINE,
        (a, b, 0, d, e, 0),
        resample=Image.Resampling.BICUBIC,
        fillcolor=0
    )

# X-axis shear comparison: 0.1, 0.3, 0.8
for factor in (0.1, 0.3, 0.8):
    result = shear_image(img, shx=factor)
    filename = f"experiment_shear_x_{factor:.1f}.jpg"
    result.save(filename, quality=95)
    print(f"Saved: {filename} | size={result.size}")
    display(result)

# Y-axis shear
y_shear_result = shear_image(img, shy=0.5)
y_shear_result.save("experiment_shear_y_0.5.jpg", quality=95)
print(f"Saved: experiment_shear_y_0.5.jpg | size={y_shear_result.size}")
display(y_shear_result)

# Combined X and Y shear
combined_shear_result = shear_image(img, shx=0.3, shy=0.2)
combined_shear_result.save("experiment_shear_xy.jpg", quality=95)
print(f"Saved: experiment_shear_xy.jpg | size={combined_shear_result.size}")
display(combined_shear_result)
```

# Intensity Transformations
Negative · Log · Power Law (Gamma)

```python
# ── 1. Negative ──────────────────────────────────── 
# Method 1: NumPy array manipulation

arr = np.asarray(img, dtype=np.uint8)
negative_arr = 255 - arr
negative_numpy = Image.fromarray(negative_arr)

negative_numpy.save("task2_1_negative_numpy.jpg", quality=95)

print("Saved: task2_1_negative_numpy.jpg")
display(negative_numpy)
```

```python
# Method 2: PIL's ImageOps

negative_pil = ImageOps.invert(img)
negative_pil.save("task2_1_negative_pil.jpg", quality=95)

print("Saved: task2_1_negative_pil.jpg")
print(
    "Both negative methods produce the same pixel values:",
    np.array_equal(np.asarray(negative_numpy), np.asarray(negative_pil))
)
display(negative_pil)
```

```python
# ── 2. Log transformation ──────────────────────────

# s = c · log(1 + r) --> We need to find c.
# We want the maximum output value (s) to be 255 when the maximum input value (r) is 255:
# 255 = c · log(1 + 255)
# c = 255 / log(1 + 255)

c = 255 / np.log1p(255)

# Convert to float BEFORE applying log1p to avoid uint8 overflow.
arr_float = np.asarray(img, dtype=np.float32)

# Apply the log transformation to each pixel
log_transformed_arr = c * np.log1p(arr_float)
log_transformed_arr = np.clip(log_transformed_arr, 0, 255).astype(np.uint8)
log_transformed_img = Image.fromarray(log_transformed_arr)

log_transformed_img.save("task2_2_log.jpg", quality=95)

print(f"c = {c:.4f}")
print("Saved: task2_2_log.jpg")
display(log_transformed_img)
```

```python
# ── 3. Power-law / Gamma correction ───────────────

# s = 255 * (r / 255)^gamma
# gamma < 1 brightens the image.
gamma = 0.5

normalized = np.asarray(img, dtype=np.float32) / 255.0
gamma_arr = 255 * np.power(normalized, gamma)
gamma_arr = np.clip(gamma_arr, 0, 255).astype(np.uint8)
gamma_img = Image.fromarray(gamma_arr)

gamma_img.save("task2_3_gamma_0.5.jpg", quality=95)

print(f"Gamma = {gamma}")
print("Saved: task2_3_gamma_0.5.jpg")
display(gamma_img)
```

```python
# Compare with gamma > 1 (darkens the image)

gamma_dark = 2.0
gamma_dark_arr = 255 * np.power(normalized, gamma_dark)
gamma_dark_arr = np.clip(gamma_dark_arr, 0, 255).astype(np.uint8)
gamma_dark_img = Image.fromarray(gamma_dark_arr)

gamma_dark_img.save("task2_3_gamma_2.0.jpg", quality=95)

print(f"Gamma = {gamma_dark}")
print("Saved: task2_3_gamma_2.0.jpg")
print("\nComparison:")
print("- gamma < 1  → brighter image")
print("- gamma = 1  → unchanged image")
print("- gamma > 1  → darker image")

display(img)
display(gamma_img)
display(gamma_dark_img)
```
